````python
# @title Check your work
# @markdown Run this cell last. If you change anything above, run the cells again from the top first.
# This cell only reads your variables. It doesn't change anything.
def check_your_work():
    import os
    found = globals()
    results = []

    upload_rows = found.get("upload_rows")
    if upload_rows is None:
        results.append((False, "upload_rows isn't defined. Upload the file with the Files panel and run the cell that reads it."))
    elif not isinstance(upload_rows, list) or not all(isinstance(row, dict) for row in upload_rows):
        results.append((False, "upload_rows should be a list of dictionaries, one per row: list(csv.DictReader(file))."))
    elif len(upload_rows) == 1724:
        results.append((True, "upload_rows has all 1,724 rows of the uploaded file."))
    else:
        results.append((False, f"upload_rows has {len(upload_rows)} rows, but sessions_export_2024_02.csv has 1,724. Check that you uploaded the February file and read it with csv.DictReader."))

    path = found.get("DRIVE_PATH")
    if not isinstance(path, str):
        results.append((False, "DRIVE_PATH isn't defined. Set it to the file's path in Drive, copied from the Files panel."))
    elif path.startswith("/content/drive/My Drive"):
        results.append((False, f"DRIVE_PATH is {path!r}. On the runtime, your Drive is /content/drive/MyDrive, without a space. Copy the path from the Files panel instead of typing it."))
    elif not path.startswith("/content/drive/"):
        results.append((False, f"DRIVE_PATH is {path!r}, which isn't in your Drive. Mount Drive, then copy the path of the file in drive/MyDrive/simmer-python from the Files panel. It starts with /content/drive/."))
    elif not os.path.exists(path):
        results.append((False, f"There's no file at {path!r}. Check that Drive is mounted and that the file is in the simmer-python folder, then copy its path from the Files panel."))
    else:
        results.append((True, "DRIVE_PATH points to the file in your Drive."))

    rows = found.get("rows")
    if rows is None:
        results.append((False, "rows isn't defined. Run the cell that reads the file from DRIVE_PATH."))
    elif isinstance(upload_rows, list) and rows == upload_rows and len(rows) == 1724:
        results.append((True, "rows has the same 1,724 rows as the uploaded file."))
    else:
        results.append((False, "rows should have the same 1,724 rows as upload_rows. Read the file from DRIVE_PATH with csv.DictReader, the same way as the uploaded file."))

    total = found.get("total_sessions")
    if total is None:
        results.append((False, "total_sessions isn't defined."))
    elif isinstance(total, str):
        results.append((False, f"total_sessions is the text {total[:20]!r}. Convert each value with int() before you add it."))
    elif total == 84654 and isinstance(total, int):
        results.append((True, "total_sessions is 84,654."))
    elif total == 84654:
        results.append((False, "total_sessions has the right value, but it should be an integer. Use int(), not float()."))
    else:
        results.append((False, f"total_sessions should be 84,654, but it is {total!r}. Add int(row[\"sessions\"]) for every row in rows."))

    email_rows = found.get("email_rows")
    if email_rows is None:
        results.append((False, "email_rows isn't defined."))
    elif not isinstance(email_rows, list):
        results.append((False, "email_rows should be a list of rows."))
    elif len(email_rows) == 112 and all(row.get("session_medium") == "email" for row in email_rows):
        results.append((True, "email_rows has the 112 email rows."))
    elif len(email_rows) == 0:
        results.append((False, "email_rows is empty. The medium is written in lowercase in the file: row[\"session_medium\"] == \"email\"."))
    else:
        results.append((False, f"email_rows has {len(email_rows)} rows, but 112 rows have the medium email. Keep the rows where row[\"session_medium\"] == \"email\"."))

    email_sessions = found.get("email_sessions")
    if email_sessions is None:
        results.append((False, "email_sessions isn't defined."))
    elif email_sessions == 5916 and isinstance(email_sessions, int):
        results.append((True, "email_sessions is 5,916."))
    elif email_sessions == 112:
        results.append((False, "email_sessions is 112: that's the number of email rows. Add up their sessions instead."))
    else:
        results.append((False, f"email_sessions should be 5,916, but it is {email_sessions!r}. Add int(row[\"sessions\"]) for every row in email_rows."))

    for passed, message in results:
        print(("✅ " if passed else "❌ ") + message)
    if all(passed for passed, message in results):
        print("\nEverything matches.")
    else:
        print("\nFix the first ❌, run the cells above again from the top, then run this cell again.")


check_your_work()
````
