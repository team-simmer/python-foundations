````python
# @title Check your work
# @markdown Run this cell last. If you change anything above, run the cells again from the top first.
# This cell only reads your functions and variables. It doesn't change anything.
def check_your_work():
    import sys
    found = globals()
    results = []

    if "tldextract" in sys.modules:
        results.append((True, "tldextract is installed and imported."))
    else:
        results.append((False, "tldextract isn't imported. Install it with %pip install tldextract in the first code cell, then import tldextract."))

    site_name = found.get("site_name")
    if not callable(site_name):
        results.append((False, "site_name() isn't defined. Run the cell that defines it."))
    else:
        cases = [
            ("out.reddit.com", "reddit"),
            ("www.pinterest.co.uk", "pinterest"),
            ("pinterest.com", "pinterest"),
            ("L.Instagram.com", "instagram"),
            ("news.google.co.jp", "google"),
            ("shop.example.com.au", "example"),
        ]
        wrong = []
        for hostname, expected in cases:
            try:
                got = site_name(hostname)
            except Exception as error:
                wrong.append((hostname, expected, f"{type(error).__name__}: {error}"))
                continue
            if got != expected:
                wrong.append((hostname, expected, got))
        if not wrong:
            results.append((True, "site_name() returns the site's name for every hostname."))
        else:
            hostname, expected, got = wrong[0]
            message = f"site_name({hostname!r}) should return {expected!r}, but it returned {got!r}."
            if got in ("co", "com"):
                message += " Splitting on dots doesn't know that some suffixes have two parts. Use tldextract.extract(hostname).domain."
            elif isinstance(got, str) and got.lower() == expected:
                message += " Return the name in lowercase: tldextract keeps the hostname's capitals."
            elif isinstance(got, str) and got.lower().startswith(expected + "."):
                message += " That's the name with its suffix. Use .domain, which is the name without it."
            elif got is None:
                message += " Use return, not print()."
            results.append((False, message))

    totals = found.get("sessions_by_site")
    expected = {"pinterest": 507, "instagram": 205, "facebook": 239, "reddit": 153, "linktr": 64}
    if totals is None:
        results.append((False, "sessions_by_site isn't defined. Run the cell that creates it."))
    elif not isinstance(totals, dict):
        results.append((False, f"sessions_by_site should be a dictionary of site names and sessions, but it is {totals!r}."))
    elif totals == expected:
        results.append((True, "sessions_by_site has the right total for every site."))
    elif not totals:
        results.append((False, "sessions_by_site is empty. If its cell stopped with an error, read the error, fix it and run the cell again."))
    elif "co" in totals or "com" in totals:
        results.append((False, f"sessions_by_site has the key 'co' or 'com', which isn't a site. Use site_name() with tldextract for every hostname. It is {totals!r}."))
    elif any(isinstance(key, str) and key != key.lower() for key in totals):
        results.append((False, f"sessions_by_site has names with capitals, so Instagram is counted apart from instagram. Return the name in lowercase from site_name(). It is {totals!r}."))
    elif sum(v for v in totals.values() if isinstance(v, int)) == 11:
        results.append((False, "sessions_by_site counts hostnames, not sessions. Add each hostname's sessions to its site's total."))
    elif sum(v for v in totals.values() if isinstance(v, int)) < 1168:
        results.append((False, f"sessions_by_site adds up to fewer than the 1,168 sessions in referrals. Add each hostname's sessions to its site's total with sessions_by_site.get(site, 0) + sessions, instead of replacing the total. It is {totals!r}."))
    else:
        results.append((False, f"sessions_by_site should be {expected}, but it is {totals!r}."))

    for passed, message in results:
        print(("✅ " if passed else "❌ ") + message)
    if all(passed for passed, message in results):
        print("\nEverything matches. Now disconnect and delete the runtime, run all, and check that you get the same result.")
    else:
        print("\nFix the first ❌, run the cells above again from the top, then run this cell again.")


check_your_work()
````
